```python
import pandas as pd

# Load data
df = pd.read_csv("/content/advertising_2023.csv")

# Validation function
def validate_advertising_data(df):
    report = {}

    report['missing_values'] = df.isnull().sum().to_dict()
    report['duplicates'] = int(df.duplicated().sum())
    report['negative_values'] = int((df < 0).any(axis=1).sum())
    report['rows'] = df.shape[0]
    report['columns'] = list(df.columns)

    return report

# Run validation
validate_advertising_data(df)
```
